```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Data copy
data = data.copy()

# Column names clean
data.columns = data.columns.str.strip()

# 2. Target variable
y = data['Churn'].astype(str).str.strip().map({
    'Yes': 1,
    'No': 0
})

# 3. Numerical columns
numeric_data = data[
    ['SeniorCitizen', 'tenure', 'MonthlyCharges', 'TotalCharges']
].copy()

# Convert TotalCharges into numeric
numeric_data['TotalCharges'] = pd.to_numeric(
    numeric_data['TotalCharges'],
    errors='coerce'
)

# 4. Categorical columns
categorical_data = data[
    ['gender', 'Partner', 'Dependents', 'Contract', 'PaymentMethod']
].copy()

# Clean categorical values
categorical_data = categorical_data.apply(
    lambda col: col.astype(str).str.strip()
)

# 5. OneHotEncoder
try:
    ohe = OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    )
except TypeError:
    ohe = OneHotEncoder(
        handle_unknown='ignore',
        sparse=False
    )

encoded = ohe.fit_transform(categorical_data)

# 6. Encoded DataFrame
encoded_categories = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(),
    index=data.index
)

# 7. Combine numerical and encoded data
df = pd.concat(
    [numeric_data, encoded_categories],
    axis=1
)

df = df.astype(float)

# 8. Independent features
X = df.copy()

# 9. Final verification
print("Features (X) Shape:", X.shape)
print("Target (y) Shape:", y.shape)

print("\nFeatures Preview:")
display(X.head())

print("\nTarget Preview:")
display(y.head())

print("\nMissing Values in X:")
print(X.isnull().sum().sum())

print("\nMissing Values in y:")
print(y.isnull().sum())

print("\nTarget Distribution:")
print(y.value_counts())

print("\nFeature Engineering Completed Successfully!")
```
